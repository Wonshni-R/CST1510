### D6. `self` is whichever object made the call

1. Copy your class from D5 and add a method `status(self)` that returns `"OVER LIMIT"` (100% or more),
   `"WARNING"` (90% or more) or `"OK"`, worked out from `self.value` and `self.limit`.
2. Create three records and print each one's `.status()`:

| Record | Expected output |
|---|---|
| `Record("R-001", 87.0, 100.0)` | `OK` |
| `Record("R-002", 120.0, 100.0)` | `OVER LIMIT` |
| `Record("R-003", 95.0, 100.0)` | `WARNING` |

````python
# D6
class Record:
    """Represent a record with a label, value and limit."""

    def __init__(self, label, value, limit):
        """Initialise the record's attributes."""
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        """Return a formatted summary of the record."""
        return f"{self.label}: {self.value}/{self.limit}"

    def status(self):                                    
        """Return the status based on the percentage."""
        percent = (self.value / self.limit) * 100    
         
        if percent >= 100:
            return "OVER LIMIT"
        elif percent >= 90:
            return "WARNING"
        else:
            return "OK"

# Create three Record objects with labels, values and limits.
r1 = Record("R-001", 87.0, 100.0)
r2 = Record("R-002", 120.0, 100.0)
r3 = Record("R-003", 95.0, 100.0)

print(r1.status())
print(r2.status())
print(r3.status())
````

### D7. One method calls another

1. Copy your class from D6 and add a method `percent(self)` that returns `(self.value / self.limit) * 100`.
2. Change `status(self)` so it uses `self.percent()` instead of working out the percentage itself.
3. Create `Record("R-001", 87.0, 100.0)` and print its `.percent()` and `.status()`.

**Expected output:** `87.0`, then `OK`

````python
# D7
class Record:
    """Represent a record with a label, value and limit."""

    def __init__(self, label, value, limit):
        """Initialise the record's attributes."""
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        """Return a formatted summary of the record."""
        return f"{self.label}: {self.value}/{self.limit}"

    def percent(self):
        """Return the value as a percentage of the limit."""
        return (self.value / self.limit) * 100

    # Determine the status using the calculated percentage.
    def status(self):
        """Return the status using the percent method."""
        p = self.percent()       # calling another method on the same object
        if p >= 100:
            return "OVER LIMIT"
        elif p >= 90:
            return "WARNING"
        else:
            return "OK"
        
r1 = Record("R-001", 87.0, 100.0)

print(r1.percent(), r1.status())
````

### D8. A method can change its own object

1. Define a class `Checker`. In `__init__(self)`, set `self.over_limit_count = 0`.
2. Add a method `check(self, value, limit)` that adds 1 to `self.over_limit_count` when `value` is
   greater than `limit`.
3. Create one `Checker`, call `.check()` three times with `(87, 100)`, `(120, 100)` and `(130, 100)`,
   then print `over_limit_count`.

**Expected output:** `2`

````python
# D8
class Checker:
    """Count values that exceed their limits."""

    def __init__(self):
        """start the over-limit count at 0."""
        self.over_limit_count = 0

    def check(self, value, limit):
        """Increase the count if the value exceeds the limit."""
        if value > limit:
            self.over_limit_count += 1    # Update and store the new total.

c = Checker()  # Create one checker object.

c.check(87, 100)
c.check(120, 100)
c.check(130, 100)

print(c.over_limit_count)
````

### D10. A class-level counter

1. Define a new, small class `Counted`. Directly under the `class` line (not inside `__init__`),

Add a comment explaining why the count is shared by every object instead of each object having its own.

````python
# D10
class Counted:
    """Count the number of objects created."""

    count = 0  # A class attribute shared by every Counted object.

    def __init__(self):
        """Increase the shared counter whenever an object is created."""
        Counted.count +=1

# Create three Counted objects.
Counted1 = Counted()
Counted2 = Counted()
Counted3 = Counted()

print(Counted.count)


# The count is shared because it is a class attribute, not an instance attribute.
````

### D11. Load objects from a file

`records_sample.txt` (in this `LAB` folder) has one record per line, like `R-001,87,100`.

1. Write `load_records(path)` that reads the file and returns a list of `Record` objects
   (convert `value` and `limit` with `float()`).
2. Loop over the list and print each record's label and status.

**Expected output:** the same three lines as D9.

````python
# D11
def load_records(path):
        """Read label,value,limit lines from a file and return a list of Records."""
        records = []

        # Read the file one line at a time.
        with open(path, "r") as f:
            for line in f:
                 label, value, limit = line.strip().split(",")

                # Convert numeric fields and create a Record object.
                 record = Record(label.strip(), float(value), float(limit))
                 records.append(record)
        return records

records = load_records("records_sample.txt")   # Load the records from the sample file.

for i in range(len(records)):
    print(records[i].label, records[i].status())
            
````

### C1. Count the records over the limit

Add a method `is_over(self)` to `Record` that returns `True` or `False`. Load `records_sample.txt` with
`load_records` from D11, use `is_over()` to count how many records are over their limit, and print a
one-line summary.

**Expected output:** `1 of 3 records are OVER LIMIT`

````python
# C1
class Record:
    """Represent a record with a label, value and limit."""

    def __init__(self, label, value, limit):
        """Initialise the record's attributes."""
        self.label = label    # Store the label in the object's attribute.
        self.value = value
        self.limit = limit

    def summary(self):
        """Return a formatted summary of the record."""
        return f"{self.label}: {self.value}/{self.limit}"

    def percent(self):
        """Calculate the percentage."""
        return (self.value / self.limit) * 100

    def status(self):
        """Return the status based on the percentage."""
        p = self.percent()        # calling another method on the same object
        if p >= 100:
            return "OVER LIMIT"
        elif p >= 90:
            return "WARNING"
        else:
            return "OK"

    def is_over(self):
        """Return True if the value exceeds the limit."""
        return self.value > self.limit         # Check whether the value has reached or exceeded the limit.

    def load_records(path):
        """Load records from a file."""
        records = []

        with open(path, "r") as f:
            for line in f:
                label, value, limit = line.strip().split(",")
                record = Record(label.strip(), float(value), float(limit))
                records.append(record)
        return records

records = load_records("records_sample.txt") 

over_limit_count = 0

for i in range(len(records)):
    if records[i].is_over():
        over_limit_count +=1  # Update and store the new total.

print(f"{over_limit_count} of {len(records)} records are OVER LIMIT")
````

### C2. An account that locks

Write a class `Account` with a `username` and a count of failed logins that starts at 0.
It has two methods:

- `fail(self)`: adds 1 to the failed logins
- `locked(self)`: returns `True` if there have been 3 or more failed logins

**Test it:**

| Code | Expected output |
|---|---|
| an account for `"anna"` with `.fail()` called 3 times, then `print(anna.locked())` | `True` |
| an account for `"ben"` with `.fail()` called once, then `print(ben.locked())` | `False` |

````python
# C2
class Account:
    """Represent an account and track failed login attempts."""

    def __init__(self, username):
       """Initialise the username and failed login count"""
       self.username = username
       self.failed_logins = 0

    def fail(self):
        """Increase the number of failed login attempts."""
        self.failed_logins +=1   # Increment failed login count.

    def locked(self):
        """Return True if there have been three or more failed logins."""
        return self.failed_logins >= 3  # Check if 3 or more logins failed.

anna = Account("anna")   # Create Anna's account and record three failed logins.
anna.fail()
anna.fail()
anna.fail()
print(anna.locked())

ben = Account("ben")  
ben.fail()
print(ben.locked())
````
